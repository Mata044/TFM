```python
from sage.rings.integer import GCD_list #Funcio que retorna el màxim comú divisor d'una llista d'enters
R.<x>=PolynomialRing(Rationals())
maxp=60 #Fixem el primer més gran que mirarem
T=prime_range(5,maxp+1) #El T que hem definit aquest matí, començant per 5
for a in range(2,maxp): #Amb les parelles (a,b) on a>b i p>a aconseguim totes les possibles corbes a considerar mòdul p
    for b in range (1,a):
        C = HyperellipticCurve(x*(x^2-a^2)*(x^2-b^2))
        L=[]
        for p in T: #Reduir la corba modul p, pels p que no divideixen el discriminant
            if not p.divides(2*a*b*(a+b)*(a-b)):
                L.append((len(C.change_ring(GF(p)).points())^2+len(C.change_ring(GF(p*p)).points()))/2-p) #Nombre de punts de la Jacobiana
            if GCD_list(L)!=16: #Comprovem si ja tenim prou informació per concloure, altrament seguim iterant
                exit
        if GCD_list(L)!=16:
            print("a =",a, ", b =",b,", m.c.d =",GCD_list(L))
```
